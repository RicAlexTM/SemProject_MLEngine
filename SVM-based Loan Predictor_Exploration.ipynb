```python
# Cell 1: Import required libraries and load the dataset file
import pandas as pd

# Load the dataset file into a pandas DataFrame
df  = pd.read_csv("SVM-based Loan Predictor.csv")

df.head()
```

```python
# Question 3a: How many rows and columns are contained in the dataset
num_rows, num_cols = df.shape
print(f"The dataset contains {num_rows} rows and {num_cols} columns.")
```

```python
# Question 3b: What datatypes are contained in the dataset?
print("Data types contained in each column of the dataset:\n")
print(df.dtypes)
```

```python
# Question 3c: Is the dataset complete? i.e.. no missing values?
total_missing = df.isnull().sum().sum()

if total_missing == 0:
    print("Yes, the dataset is complete. There are no missing values.")
else:
    print(f"No, the dataset is not complete. It contains {total_missing} missing values(s) in total \n")
    print("Missing values per column:")
    print(df.isnull().sum()[df.isnull().sum() > 0])
```

```python
# Question 3d: Slice out the first 15 rows and last 20 rows, merge into "df_sample", and display.

# Slice first 15 rows
first_15 = df.head(15)

# Slice last 20 rows
last_20 = df.tail(20)

# Merge both slice into a new dataframe titled "df_sample"
df_sample = pd.concat([first_15, last_20], ignore_index=True)

# Display the merged sample dataframe
df_sample
```
